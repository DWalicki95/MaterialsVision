```python
import cv2
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
img_basepath = Path('/Users/dawidwalicki/Doktorat/images')

img_name = 'AS1A_40_5_jpg.rf.5a26ead52dd16332d7576e2a02bd44dc.jpg'
```

```python
img_fullpath = img_basepath / img_name
```

```python
img = cv2.imread(img_fullpath)
```

```python
def show_img(img):
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.imshow(img_rgb)
    plt.axis('off')
    plt.show()
```

```python
show_img(img=img)
```

## Top Hat

```python
filterSize =(24, 24)
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, filterSize)
```

```python
tophat_img = cv2.morphologyEx(img, cv2.MORPH_TOPHAT, kernel)
```

```python
show_img(tophat_img)
```

## Krok po kroku

```python
img_step1 = (img // 4)
```

```python
show_img(img_step1)
```
